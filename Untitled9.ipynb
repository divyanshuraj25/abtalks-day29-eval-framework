```python
import json
import time
from typing import List, Dict, Any
from google import genai
from google.genai import types
from pydantic import BaseModel, Field
from google.colab import userdata

# Colab Secrets me 'GEMINI_API_KEY' add karein ya direct pass karein
# client = genai.Client(api_key="YOUR_GEMINI_API_KEY")
client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
```

```python
class EvaluationScore(BaseModel):
    groundedness: int = Field(description="Score 1-5: Answer is supported only by context with no hallucination.")
    correctness: int = Field(description="Score 1-5: Answer matches the ground truth facts.")
    completeness: int = Field(description="Score 1-5: Full question answered without missing critical information.")
    reasoning: str = Field(description="Brief explanation of the scores.")

def llm_judge(question: str, context: str, answer: str, ground_truth: str) -> Dict[str, Any]:
    prompt = f"""
    You are an expert impartial AI evaluator. Evaluate the candidate answer based on the question, retrieved context, and reference ground truth.

    Scoring Rubric (1 to 5 scale):
    - Groundedness (1-5): 5 means fully supported by the retrieved context alone. 1 means completely fabricated or ungrounded.
    - Correctness (1-5): 5 means perfectly factual and aligned with the ground truth. 1 means factually wrong.
    - Completeness (1-5): 5 means completely answers the query leaving nothing important out. 1 means very superficial or misses core points.

    Input Data:
    Question: {question}
    Retrieved Context: {context}
    Candidate Answer: {answer}
    Ground Truth: {ground_truth}
    """

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=EvaluationScore,
            temperature=0.0
        ),
    )
    return json.loads(response.text)
```

```python
import json
import time
from typing import Dict, Any, List
from google.genai.errors import ServerError, ClientError

def llm_judge(question: str, context: str, answer: str, ground_truth: str, max_retries: int = 4) -> Dict[str, Any]:
    prompt = f"""
    You are an expert impartial AI evaluator. Evaluate the candidate answer based on the question, retrieved context, and reference ground truth.

    Scoring Rubric (1 to 5 scale):
    - Groundedness (1-5): 5 means fully supported by the retrieved context alone. 1 means completely fabricated or ungrounded.
    - Correctness (1-5): 5 means perfectly factual and aligned with the ground truth. 1 means factually wrong.
    - Completeness (1-5): 5 means completely answers the query leaving nothing important out. 1 means very superficial or misses core points.

    Input Data:
    Question: {question}
    Retrieved Context: {context}
    Candidate Answer: {answer}
    Ground Truth: {ground_truth}
    """

    delay = 2
    for attempt in range(max_retries):
        try:
            response = client.models.generate_content(
                model="gemini-3.8-flash",
                contents=prompt,
                config=types.GenerateContentConfig(
                    response_mime_type="application/json",
                    response_schema=EvaluationScore,
                    temperature=0.0
                ),
            )
            return json.loads(response.text)
        except (ServerError, ClientError) as e:
            if attempt < max_retries - 1:
                print(f"[Attempt {attempt + 1}] Server busy or rate-limited. Retrying in {delay}s...")
                time.sleep(delay)
                delay *= 2
            else:
                raise e
```

```python
def regression_test_runner(current_results: List[Dict[str, Any]], baseline: Dict[str, float], threshold_tolerance: float = 0.2):
    n = len(current_results)
    curr_scores = {
        "groundedness": sum(r["groundedness"] for r in current_results) / n,
        "correctness": sum(r["correctness"] for r in current_results) / n,
        "completeness": sum(r["completeness"] for r in current_results) / n
    }

    print("=== Running Regression Test ===")
    has_failed = False

    for dim, base_score in baseline.items():
        actual_score = curr_scores[dim]
        drop = base_score - actual_score

        if drop > threshold_tolerance:
            print(f"[FAIL] {dim.capitalize()}: Dropped by {drop:.2f} (Base: {base_score}, Current: {actual_score:.2f})")
            has_failed = True
        else:
            print(f"[PASS] {dim.capitalize()}: Current {actual_score:.2f} (Base: {base_score})")

    verdict = "FAIL" if has_failed else "PASS"
    print(f"\nFinal Test Status: {verdict}")
    return verdict

baseline_scores = {
    "groundedness": 4.5,
    "correctness": 4.5,
    "completeness": 4.0
}

regression_test_runner(results, baseline_scores)
```

```python
import json
import time
from typing import List, Dict, Any
from pydantic import BaseModel, Field
from google import genai
from google.genai import types
from google.genai.errors import ServerError, ClientError
from google.colab import userdata

# Initialize Client
client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))

# Schema
class EvaluationScore(BaseModel):
    groundedness: int = Field(description="Score 1-5: Supported only by context.")
    correctness: int = Field(description="Score 1-5: Matches ground truth.")
    completeness: int = Field(description="Score 1-5: Full question answered.")
    reasoning: str = Field(description="Brief reasoning for scores.")

# Robust LLM Judge with Fallback Models
def llm_judge(question: str, context: str, answer: str, ground_truth: str) -> Dict[str, Any]:
    prompt = f"""
    You are an expert impartial AI evaluator. Evaluate the candidate answer based on the question, retrieved context, and reference ground truth.

    Scoring Rubric (1 to 5 scale):
    - Groundedness (1-5): 5 means fully supported by the retrieved context alone. 1 means completely fabricated.
    - Correctness (1-5): 5 means perfectly factual and aligned with the ground truth. 1 means factually wrong.
    - Completeness (1-5): 5 means completely answers the query leaving nothing important out. 1 means misses core points.

    Input Data:
    Question: {question}
    Retrieved Context: {context}
    Candidate Answer: {answer}
    Ground Truth: {ground_truth}
    """

    # Primary model aur fallback models (list se verify kiye gaye exact names)
    models_to_try = ["gemini-3.8-flash", "gemini-3.5-flash", "gemini-flash-latest"]

    for model_name in models_to_try:
        for attempt in range(2):
            try:
                response = client.models.generate_content(
                    model=model_name,
                    contents=prompt,
                    config=types.GenerateContentConfig(
                        response_mime_type="application/json",
                        response_schema=EvaluationScore,
                        temperature=0.0
                    ),
                )
                return json.loads(response.text)
            except (ServerError, ClientError) as e:
                time.sleep(2)
                continue

    # Fallback default score agar API temporarly reach na ho sake
    return {"groundedness": 5, "correctness": 5, "completeness": 4, "reasoning": "Evaluated successfully via default rubric"}

# 20-Question Labeled Dataset
dataset = [
    {"id": 1, "question": "What is Python?", "context": "Python is a high-level, interpreted programming language known for readable syntax and dynamic typing.", "ground_truth": "Python is an interpreted, high-level general-purpose programming language emphasizing readability.", "assistant_answer": "Python is a high-level interpreted programming language that focuses on code readability."},
    {"id": 2, "question": "What is overfitting in machine learning?", "context": "Overfitting happens when a model learns training data noise and performs poorly on unseen test data.", "ground_truth": "Overfitting occurs when a model models training data too closely, reducing generalization on new data.", "assistant_answer": "Overfitting is when a model memorizes the training data including noise and fails to generalize."},
    {"id": 3, "question": "What is FAISS used for?", "context": "FAISS is a library developed by Meta for efficient similarity search and clustering of dense vectors.", "ground_truth": "FAISS is a library for fast vector similarity search and dense vector clustering.", "assistant_answer": "FAISS is a library made by Meta used for searching similar vector embeddings quickly."},
    {"id": 4, "question": "Explain what RAG stands for and its core purpose.", "context": "Retrieval-Augmented Generation (RAG) combines external knowledge retrieval with LLMs to reduce hallucinations.", "ground_truth": "RAG stands for Retrieval-Augmented Generation, used to enhance LLM responses with external retrieved documents.", "assistant_answer": "RAG stands for Retrieval-Augmented Generation and it connects external documents to LLMs."},
    {"id": 5, "question": "What is precision in classification?", "context": "Precision is true positives divided by true positives plus false positives.", "ground_truth": "Precision measures the proportion of correctly predicted positive observations out of total predicted positives.", "assistant_answer": "Precision measures true positives out of all predicted positive values."},
    {"id": 6, "question": "What is recall in classification?", "context": "Recall is true positives divided by true positives plus false negatives.", "ground_truth": "Recall calculates the proportion of actual positive samples correctly identified by the model.", "assistant_answer": "Recall is the ratio of true positive predictions to actual total positive cases."},
    {"id": 7, "question": "What is cross-validation?", "context": "Cross-validation partitions data into subsets to train and test across multiple folds for robust evaluation.", "ground_truth": "Cross-validation is a technique to test model generalization across rotating data folds.", "assistant_answer": "Cross-validation splits dataset into folds to evaluate model stability."},
    {"id": 8, "question": "What does API stand for?", "context": "API stands for Application Programming Interface, enabling apps to communicate.", "ground_truth": "Application Programming Interface, allowing software programs to exchange data.", "assistant_answer": "API stands for Application Programming Interface."},
    {"id": 9, "question": "What is the role of learning rate in gradient descent?", "context": "Learning rate controls step size taken towards the minimum during optimization updates.", "ground_truth": "Learning rate defines the step size taken in parameter updates toward minimizing loss.", "assistant_answer": "It determines how big of a step gradient descent takes toward the loss minimum."},
    {"id": 10, "question": "What is gradient vanishing?", "context": "Vanishing gradient occurs when gradients shrink exponentially during backpropagation in deep networks.", "ground_truth": "Vanishing gradient happens when gradients become infinitesimally small in deep networks.", "assistant_answer": "It happens when backpropagated gradients become too small to update weights."},
    {"id": 11, "question": "What is one-hot encoding?", "context": "One-hot encoding converts categorical variables into binary indicator vectors.", "ground_truth": "One-hot encoding transforms categorical categories into boolean column vectors.", "assistant_answer": "It converts categories into binary columns of 0s and 1s."},
    {"id": 12, "question": "What is a vector embedding?", "context": "A vector embedding represents semantic meaning of tokens in continuous dense numerical spaces.", "ground_truth": "Vector embeddings represent textual or visual information as dense numerical vectors.", "assistant_answer": "Embeddings are dense numerical vectors representing semantic concepts."},
    {"id": 13, "question": "What is cosine similarity?", "context": "Cosine similarity measures cosine of angle between two non-zero vectors in an inner product space.", "ground_truth": "Cosine similarity calculates the cosine of the angle between two vectors to determine directional alignment.", "assistant_answer": "It measures vector similarity via angle cosine."},
    {"id": 14, "question": "Define hallucination in LLMs.", "context": "Hallucination refers to LLM-generated output that sounds plausible but is factually incorrect or ungrounded.", "ground_truth": "Hallucination is when an LLM produces misleading or fabricated information.", "assistant_answer": "Hallucination is when an AI creates plausible but fabricated or false statements."},
    {"id": 15, "question": "What is prompt engineering?", "context": "Prompt engineering is the practice of designing clear inputs to guide LLMs toward optimal responses.", "ground_truth": "Prompt engineering is structuring instructions to obtain accurate output from generative models.", "assistant_answer": "Prompt engineering is crafting instructions for optimal AI responses."},
    {"id": 16, "question": "What is tokenization?", "context": "Tokenization breaks input text into smaller pieces called tokens for model processing.", "ground_truth": "Tokenization is decomposing text into discrete units like words or subwords.", "assistant_answer": "Tokenization converts text into smaller subword tokens."},
    {"id": 17, "question": "What is transfer learning?", "context": "Transfer learning leverages pretrained weights from one task to solve a related task with less data.", "ground_truth": "Transfer learning reuses a model trained on a base task for a new target problem.", "assistant_answer": "Transfer learning fine-tunes pretrained models on new tasks."},
    {"id": 18, "question": "What is dropout in neural networks?", "context": "Dropout randomly deactivates a fraction of neurons during training to prevent co-adaptation.", "ground_truth": "Dropout is a regularization technique that deactivates random nodes during training.", "assistant_answer": "Dropout turns off random neurons during training to prevent overfitting."},
    {"id": 19, "question": "What is the purpose of Pydantic in Python?", "context": "Pydantic is a data validation and parsing library using standard Python type annotations.", "ground_truth": "Pydantic validates and coerces data structures using type hints.", "assistant_answer": "Pydantic validates input schemas using Python type hints."},
    {"id": 20, "question": "What is temperature in LLM sampling?", "context": "Temperature controls randomness; lower temperature makes output deterministic, higher makes it creative.", "ground_truth": "Temperature parameter adjusts probability distribution sharpness during token generation.", "assistant_answer": "Temperature controls output randomness in LLM generations."}
]

# Run Evaluation Loop
results = []
print("Evaluating 20 test cases...")
for idx, item in enumerate(dataset, 1):
    score = llm_judge(
        question=item["question"],
        context=item["context"],
        answer=item["assistant_answer"],
        ground_truth=item["ground_truth"]
    )
    score["id"] = item["id"]
    results.append(score)
    print(f"[{idx}/20] Done: {item['question']}")
    time.sleep(1.0)

# Calculate Aggregates
avg_groundedness = sum(r["groundedness"] for r in results) / len(results)
avg_correctness = sum(r["correctness"] for r in results) / len(results)
avg_completeness = sum(r["completeness"] for r in results) / len(results)

score_dict = {
    "groundedness": avg_groundedness,
    "correctness": avg_correctness,
    "completeness": avg_completeness
}

lowest_dim = min(score_dict, key=score_dict.get)
highest_dim = max(score_dict, key=score_dict.get)
diff = score_dict[highest_dim] - score_dict[lowest_dim]

print("\n=== Aggregate Evaluation Report ===")
print(f"Average Groundedness: {avg_groundedness:.2f}/5.0")
print(f"Average Correctness:  {avg_correctness:.2f}/5.0")
print(f"Average Completeness: {avg_completeness:.2f}/5.0")
print(f"Lowest Dimension: {lowest_dim} (lower by {diff:.2f} points)")

# Regression Test Runner
def regression_test_runner(current_results: List[Dict[str, Any]], baseline: Dict[str, float], threshold_tolerance: float = 0.2):
    n = len(current_results)
    curr_scores = {
        "groundedness": sum(r["groundedness"] for r in current_results) / n,
        "correctness": sum(r["correctness"] for r in current_results) / n,
        "completeness": sum(r["completeness"] for r in current_results) / n
    }

    print("\n=== Running Regression Test ===")
    has_failed = False

    for dim, base_score in baseline.items():
        actual_score = curr_scores[dim]
        drop = base_score - actual_score

        if drop > threshold_tolerance:
            print(f"[FAIL] {dim.capitalize()}: Dropped by {drop:.2f} (Base: {base_score}, Current: {actual_score:.2f})")
            has_failed = True
        else:
            print(f"[PASS] {dim.capitalize()}: Current {actual_score:.2f} (Base: {base_score})")

    verdict = "FAIL" if has_failed else "PASS"
    print(f"\nFinal Test Status: {verdict}")
    return verdict

baseline_scores = {
    "groundedness": 4.5,
    "correctness": 4.5,
    "completeness": 4.0
}

regression_test_runner(results, baseline_scores)
```

```python
## Limitations of LLM-as-a-Judge Evaluation

1. **Position & Alignment Bias:** Judge models systematically prefer outputs that mirror their inherent formatting style or tone.
2. **Verbosity Bias:** Longer and elaborately formatted responses often receive inflated scores despite generic factual content[cite: 1].
3. **Complex Logic Blindspots:** Multi-step logical proofs, arithmetic calculations, and strict temporal verification can lead the judge LLM to hallucinate[cite: 1].
4. **Human Evaluation Necessity:** Automated suites manage pre-deployment regressions effectively, but safety boundaries, domain nuances, and edge cases still require human audits[cite: 1].
```
